```python
import numpy as np
from scipy import constants
from uncertainties import ufloat
from uncertainties import unumpy
```

# Connecting diodes

```python
def R(U_0, U_diode, I_max):
    return (U_0 - U_diode) / I_max 
```

```python
U_0 = 5  # V
colour = ["blue", "green (515 nm)", "green (565 nm)", "yellow", "red"]
U_diode = np.array([3.5, 3.3, 2.2, 2.0, 1.85])  # V
I_max = np.array([30, 25, 25, 25, 30]) * 1.0e-3  # A
I_betrieb = 10e-3  # A

for i in range(len(colour)):
    resistance = np.round(R(U_0, U_diode[i], I_betrieb))
    print(f"${colour[i]}:\\quad R \\geq {int(resistance)}\\,\\Omega$")
```
