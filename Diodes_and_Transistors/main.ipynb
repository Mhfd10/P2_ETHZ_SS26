```python
import numpy as np
from scipy import constants
from uncertainties import ufloat
from uncertainties import unumpy
```

# Connecting diodes

```python
def R(U_0, U_diode, I_max):
    return (U_0 - U_diode) / I_max 
```

```python
U_0 = 5  # V
colour = ["blue", "green (515 nm)", "green (565 nm)", "yellow", "red"]
U_diode = np.array([3.5, 3.3, 2.2, 2.0, 1.85])  # V
I_max = np.array([30, 25, 25, 25, 30]) * 1.0e-3  # A
I_betrieb = 10e-3  # A

for i in range(len(colour)):
    resistance = np.round(R(U_0, U_diode[i], I_betrieb))
    print(f"${colour[i]}:\\quad R \\geq {int(resistance)}\\,\\Omega$")
```

# Photoelectric effect

```python
def h(U, lmd):
    return constants.e*U*lmd/constants.c
```

```python
lambda_led = np.array([468, 515, 565, 592, 660]) * 10**(-9)  # m
sigma_lambda_led = 0.5 * 10**(-9) # m
lambda_led_u = unumpy.uarray(lambda_led, sigma_lambda_led) # uncertainty array

U_faint = np.array([3, 1.9 , 1.8, 1.8, 1.7]) # V
sigma_U_faint = 0.05 # V
U_faint_u = unumpy.uarray(U_faint, sigma_U_faint) # uncertainty array

U_10muA  = np.array([2.23, 2.053, 1.666, 1.592, 1.475]) # V
sigma_U_10muA = 0.05 # V
U_10muA_u = unumpy.uarray(U_10muA, sigma_U_10muA) # uncertainty array

# calculate h
h_faint= h(U_faint_u, lambda_led_u)
h_10uA = h(U_10muA_u, lambda_led_u)
h_total = np.concatenate((h_faint, h_10uA))
h_mean = np.mean(h_total)
h_theory = constants.h

# extract nominal values
h_faint_n = unumpy.nominal_values(h_faint)
h_10uA_n  = unumpy.nominal_values(h_10uA)
h_mean_n  = unumpy.nominal_values(h_mean)

for i in range(len(colour)):
    dev_faint_abs = h_faint_n[i] - h_theory
    dev_10uA_abs  = h_10uA_n[i] - h_theory

    dev_faint_rel = np.abs(dev_faint_abs / h_theory * 100)
    dev_10uA_rel  = np.abs(dev_10uA_abs / h_theory * 100)

    print(f"{colour[i]}:")
    print(f"h (faint) = {h_faint[i]:.2u} Js")
    print(f"rel. deviation = {dev_faint_rel:.2f} %")
    print(f"h (10 µA) = {h_10uA[i]:.2u} Js")
    print(f"rel. deviation = {dev_10uA_rel:.2f} % \n")

h_mean_abs = h_mean_n - h_theory
h_mean_rel = np.abs(h_mean_abs / h_theory * 100)

print(f"h mean = {h_mean:.2u} Js")
print(f"rel. deviation = {h_mean_rel:.2f} %")
```

# Illuminated LED as receiver?

```python
U_open = ufloat(0.870, 0.0005*10**(-3)) # V
R_multi = 10*10**(6) # Ω
P_open = U_open ** 2 / R_multi * 10**(3) # mW
print(f"estimated P open = {P_open:.2e} mW" )
U_diode = ufloat(1.178, 0.0005) # V
I_diode = ufloat(8.19, 0.005) # mA
P_diode = U_diode*I_diode
print(f"Measured Power through diode = {P_diode:.2} mW")

Power_transmitted_est = 160/(5*100) # %
Power_trans = Power_transmitted_est * P_diode
print(f"Estimated Powert transmitted = {Power_trans:.2f} mW")

Power_loss_est = np.abs(P_open - Power_trans)
print(f"estimated Power loss through transmission = {Power_loss_est:.2f} mW")
```
